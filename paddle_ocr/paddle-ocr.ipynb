```python
import json
import glob
import numpy as np
from PIL import Image
from paddleocr import PaddleOCR
```

```python
class pd_ocr:
    def __init__(self):
        self.ocr = PaddleOCR(
            use_textline_orientation=True, 
            lang='hi',
            engine="transformers",
            device="gpu:0",
        )

    def get_output(self, files, layout = 'fixed'):
        for file in files:
            img_array = np.array(Image.open(file).convert('RGB'))
            result = self.ocr.predict(img_array)
            filename = file.split('/')[-1]
            for res in result:
                res.save_to_img(f"output/{layout}_layout/{filename}")
                res.save_to_json(f"output/{layout}_layout/{filename.replace('.png', '.json').replace('.jpg', '.json')}")

obj = pd_ocr()
```

```python
fix_img_files = glob.glob("/kaggle/input/datasets/eddwait/km-take-home-assessment/fixed-layout/img/*.png")
var_img_files = glob.glob("/kaggle/input/datasets/eddwait/km-take-home-assessment/variable-layout/img/*.png")
```

```python
%%time
obj.get_output(fix_img_files)
```

```python
%%time
obj.get_output(var_img_files, layout = "variable")
```
